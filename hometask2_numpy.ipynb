Задача 1: Создать массив 10x10 заполненный в центре нулями, а по границе единицами.

```python
import numpy as np 
a = np.pad(np.zeros((8,8)), 1, constant_values = 1)
print(a)
```

Задача 2: Создайте массив 8х8, заполненный 0 и 1 в шахматном порядке.

```python
a = np.indices((8, 8)).sum(axis=0) % 2
print(a)
```

Задача 3: Создайте матрицу 5x5 со значениями 1,2,3,4 под главной диагональю. Остальные значения - нули

```python
a = np.diag(np.arange(1, 5), k=-1)
print(a)
```

Задача 4: Создайте случайный массив 5х5 и нормируйте его так, чтобы среднее вдоль каждой строки было равно нулю.

```python
a = np.random.rand(5, 5)
a = a - a.mean(axis=1, keepdims=True)
print(a)
```

Задача 5: Создайте двумерный массив 10x3, заполненный случайными числами от 0 до 1. В каждой строке выберите значение, наиболее близкое к 0.5.

```python
a = np.random.rand(10, 3)
result = a[np.arange(10), np.abs(a - 0.5).argmin(axis=1)]
print(result)
```

Задача 6: Создайте двумерный массив 6x6, заполненный случайными значениями. Сумму по каждой строке поделите на максимальный элемент из столбца с тем же индексом.

```python
a = np.random.rand(6, 6)
result = a.sum(axis=1) / a.max(axis=0)
print(result)
```

Задача 7: На примере массива x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) найдите максимальный элемент в массиве среди элементов, перед которыми стоит ноль.

```python
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
result = x[1:][x[:-1] == 0].max()
print(result)
```

Задача 8: Пусть заданы два массива x = np.ones(10) и i = np.array([0, 1, 2, 3, 5, 5, 5, 8]). Прибавьте единицу к тем элементам массива x, индексы которых указаны в массиве i. В случае, если некоторый индекс встретился в массиве i несколько раз, прибавьте к соответствующему элементу массива x число вхождений данного индекса в массив i.

```python
x = np.ones(10)
i = np.array([0, 1, 2, 3, 5, 5, 5, 8])

np.add.at(x, i, 1)
print(x)
```

Задача 9: Задайте два массива a = np.array([[1, 4.9, 2.4], [7.1, 7, 9.8], [2, 4.6, 1.2]]) и b = np.array([3, 5, 10, 0]). Создайте массив, в котором каждый элемент первого массива заменен на ближайший к нему элемент второго массива.

```python
a = np.array([[1, 4.9, 2.4], [7.1, 7, 9.8], [2, 4.6, 1.2]])
b = np.array([3, 5, 10, 0])

result = b[np.abs(a[:, :, None] - b).argmin(axis=2)]
print(result)
```

Задача 10: Найдите скользящие средние с окном размера 3 по массиву a = np.array([1, 4.9, 2.4, 7.1, 7, 9.8, 2, -5, 4.6, 1.2]).

```python
a = np.array([1, 4.9, 2.4, 7.1, 7, 9.8, 2, -5, 4.6, 1.2])
result = np.convolve(a, np.ones(3) / 3, mode='valid')
print(result)
```
